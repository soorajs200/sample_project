# Q1. Write a function that return the range (maximum and minimun) of a given list of numbers

```python
def min_max(num):
    x = min(num)
    y = max(num)
    print("range({},{})".format(x,y))

a = [1,2,3,4,5,6,7]
min_max(a)
```

## Q2. Write function that returns mean value of a list of numbers

```python
def mean(lst):
    sum=0
    n=len(lst)
    for a in lst:
        sum += a
    avg=sum//n
    return avg
ac=[23,45,65,12]
mean(ac)
```

# Q3. Write a function that returns the median value of a list of numbers

```python
def median(lst):
    srt = sorted(lst)
    n=len(srt)
    if n%2 == 0:
        print("the median is",(srt[(n//2)-1]+srt[n//2])//2)
    else:
        print("the median is", (srt[((n+1)//2)-1]))

ab=[4,6,8,2,9,7,5,3,1,12,16]
median(ab)
```

## Q4. Write a function to return the mod of a list of numbers, if no mod exists, it should return None. If more than one mod exist, return the mod values as list

```python
def find_mode(lst):
    freq = {}
    for num in lst:
        freq[num] = freq.get(num, 0) + 1

    max_count = max(freq.values())
    mode = [key for key, value in freq.items() if value == max_count]

    if len(mode) == len(freq):
        return "No mode"
    elif len(mode) == 1:
        return mode[0]
    else:
        return mode
find_mode(ab)
```

## Q5. Write function to compute a given percentile value (e.g. 90th percentile) of a list of numbers

```python
def percentile(lst,p):
    lst.sort()
    n=len(lst)
    g=float((p/100)*(n+1))
    #print("The value is",g-1)
    if g%10==0:
        print(lst[g-1])
    else:
        f=int(g)
        h=g-f
        #print("The numbers are ",lst[f-1],",",lst[f])
        num = lst[f-1]+(lst[f]-lst[f-1])*h
        #print("The number is ",num)
        return num

percentile(ab,50)
```

# Q6. Write a function that returns the lower bound of a list of numbers

```python
gh=[4,6,8,2,9,7,5,3]
def lower_bound(lst):
  q1=percentile(lst,25)
  q2=percentile(lst,50)
  q3=percentile(lst,75)
  IQR=q3-q1
  LB=q1-1.5*IQR
  return LB

lower_bound(gh)

```

## Q7. Write function to return any outliers from a given list of numbers, if no outliers present it should return None

```python
def upper_bound(lst):
  q1=percentile(lst,25)
  q2=percentile(lst,50)
  q3=percentile(lst,75)
  IQR=q3-q1
  UB=q3+1.5*IQR
  return UB
upper_bound(gh)

def outlier(lst):
  lst.sort()
  ub=upper_bound(lst)
  lb=lower_bound(lst)
  out=[]
  for i in lst:
    if i < lb or i > ub:
      out.append(i)
  if not out:
    print("no outliers")
  else:
    print(out)
tg= [10, 12, 13, 12, 11, 14, 15, 16, 100, 105]
outlier(tg)
```

##Q8. Write a function to return standard deviation of a list of numbers

```python
def std_dev(lst):
  n=len(lst)
  mean=sum(lst)//n
  tot=0
  for i in lst:
    tot=tot+((i-mean)**2)
  var=float(tot/n)
  sd=var**0.5
  return sd
std_dev(tg)
```

## Q9. Write function to compute and return Z-score values of a list of numbers

```python
def z_score(lst):
  n=len(lst)
  mean=sum(lst)//n
  tot=0
  for i in lst:
    tot=tot+((i-mean)**2)
  var=float(tot/n)
  sd=var**0.5
  z=[]
  for i in lst:
    z.append((i-mean)/sd)
  return z
hg=[2,2,2,-2,-2,-2]
z_score(hg)

```

## Q10. Write a function to compute and return the Pearson correlation coefficient value of two list of numbers

```python
def pearson_corr(x, y):
    if len(x) != len(y):
        raise ValueError("Both lists must have the same length.")

    n = len(x)
    mean_x = sum(x) / n
    mean_y = sum(y) / n

    num = sum((x[i] - mean_x) * (y[i] - mean_y) for i in range(n))
    den_x = sum((x[i] - mean_x) ** 2 for i in range(n)) ** 0.5
    den_y = sum((y[i] - mean_y) ** 2 for i in range(n)) ** 0.5

    if den_x == 0 or den_y == 0:
        return 0  # No variation → correlation undefined

    r = num / (den_x * den_y)
    return r
pearson_corr(hg,tg)

```
